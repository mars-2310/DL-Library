```python
import activations as act
```

```python
def neuron(x, w, b, activation):
    x, w, b = np.asarray(x), np.asarray(w), np.asarray(b)
    res =  (np.dot(w, x.T) + b)
    return np.asarray(activation(res), dtype=np.float32)
```

```python
w = np.array([78, 89])
x = np.array([7.1, 5.1])
b = 0.0

neuron(x, w, b, act.sigmoid)
```
